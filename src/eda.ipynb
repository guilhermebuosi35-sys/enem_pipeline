## Interpretações


- A tabela de Resultados possui 2.767.147 valores NULLs referente as informações de escolaridade dos candidatos. Podemos inferir que o dataset contenha:

    1. Candidatos que já concluíram o ensino médio há anos e a escola não está mais ativa ou não tem registro no sistema do INEP. O ENEM não é só para quem está na escola, e qualquer pessoa com mais de 18 anos pode fazer;
    2. Candidatos que estudaram em escolas no exterior ou em sistemas de ensino não cadastrados no INEP;
    3. Treineiros: alunos que fazem o ENEM antes de concluir o ensino médio como treino. Eles podem ter escola mas o vínculo não estar preenchido;
    4. Problemas de cadastro: o participante não preencheu ou preencheu errado no momento da inscrição;

- Há uma variação na existência de valores na presença de candidatos entre os quatro âmbitos de prova (Redação + Linguagens, Matemática, Ciências Humanas e Ciências da Natureza). Comum visto que candidatos podem faltar em um dos dois dias de prova, ou ir em apenas um deles.
- As tabelas não possuem chave em comum conforme documentado no dicionário de dados do INEP, portanto serão tratadas e carregadas separadamente no banco.
- A tabela com as informações dos participantes possui poucos valores nulos para serem tratados;

    TB_ENSINO é a principal, onde contém o Tipo de instituição que o participante concluiu ou concluirá o Ensino Médio (fonte fácil de não ser colhida) 

    ```
    TP_ENSINO    2787062
    6               9
    8               1
    9               1
    6               1
    9               2
    ```
- A tabela de `participantes` possui um questionário socioecônomico que permite aprofundar as analises de desempenho de acordo com cada cenário social do participante;
- A grande maioria dos dados são integers ou floats — Isso significa que a base usa de códigos ENUM para representar os dados em boa parte dos atributos, exigindo estrutura de condições para facilitar a leitura;
- A base `itens_prova` contém dados sobre as questões dadas dentro da prova, o que pode nos dar informações sobre as dificuldades das questões e a influência sobre o desempenho dos participantes sobre elas.
